```python
import pandas as pd
import numpy as np


# abrimos el archivo usando una función específica de pandas
raw_dataset = pd.read_csv("CKD_NHANES_2021_2023.csv")

print(raw_dataset)
```

```python

# 1. Define tu lista personalizable de valores nulos (¡modifícala a tu gusto!)
valores_nulos = ["unknown", "Unknown", "UNKNOWN", "n/a", "NA", "?", "","7","9",7,9]

# 2. Reemplazamos esos valores por np.nan en una copia temporal
# Usamos regex=False para evitar problemas con caracteres especiales
df_temp = raw_dataset.replace(valores_nulos, np.nan)

# 3. Calculamos el porcentaje de nulos por columna (variable)
porcentaje_nulos = (df_temp.isnull().mean() * 100).round(2)

# 4. Mostramos el resultado ordenado de mayor a menor porcentaje
print(porcentaje_nulos.sort_values(ascending=False))
```

```python
columna = 'ckd_stage'

# Muestra todos los valores únicos y su frecuencia (incluyendo NaN)
print(raw_dataset[columna].value_counts(dropna=False))
```

```python
#Calculo los nulos ahora si sin considerar los unknown
porcentaje_nulos = (raw_dataset.isnull().mean() * 100).round(2)

print(porcentaje_nulos.sort_values(ascending=False))
```

```python
import numpy as np
import scipy.stats as stats

columna = 'education_level'

# Extraemos la columna limpia (sin NaN ni valores vacíos)
datos = raw_dataset[columna].dropna()
```

```python
from scipy.stats import shapiro

# Separamos los datos en dos grupos, desarrollados y en vías de desarrollo
NegrosCKD = raw_dataset[raw_dataset['ethnicity'] == "Non-Hispanic White"],['ckd_present']

# Test de Shapiro-Wilk para países desarrollados
stat, p = shapiro(NegrosCKD)
print(f"Test de Shapiro-Wilk para países desarrollados: Estadístico={stat:.3f}, p-valor={p:.3f}")

```

```python
import matplotlib.pyplot as plt
import scipy.stats as stats

# QQ plot para países desarrollados
stats.probplot(NegrosCKD, dist="norm", plot=plt)
plt.title("Losnegros")
plt.show()
```
